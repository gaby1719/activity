```python
limite_inferior = 0
limite_superior = 0
variables_numericas = ["cantidad"]

for columna in variables_numericas:
    media = df[columna].mean()
    desviacion_estandar = df[columna].std()

    limite_inferior = media - 2 * desviacion_estandar
    limite_superior = media + 2 * desviacion_estandar

    outliers = df[
        (df[columna] < limite_inferior) |
        (df[columna] > limite_superior)
    ]

    print("Variable:", columna)
    print("Media:", round(media, 2))
    print("Desviación estándar:", round(desviacion_estandar, 2))
    print("Límite inferior:", round(limite_inferior, 2))
    print("Límite superior:", round(limite_superior, 2))
    print("Cantidad de valores atípicos:", len(outliers))
    print("-" * 40)
```

```python
columnas_asociacion = [
    "transaccion_id",
    "cliente_id",
    "fecha",
    "categoria",
    "item",
    "cantidad",
    "canal"
]
```

```python
df_asociacion = df[columnas_asociacion].copy()

df_asociacion.head()
```

```python
for columna in columnas_asociacion:
    print(columna, df_asociacion[columna].unique())
```

```python
df_asociacion_bool = df_asociacion.astype(bool)

df_asociacion_bool.head()
```

```python
df_asociacion_bool.dtypes
```

```python
df_asociacion_encoded = pd.crosstab(df_asociacion['transaccion_id'], df_asociacion['item'])
frecuencias = df_asociacion_encoded.sum().sort_values(ascending=False)
frecuencias
```

```python
total_transacciones = df_asociacion_encoded.shape[0]
porcentajes = (frecuencias / total_transacciones * 100).sort_values(ascending=False)

porcentajes.round(2)
```

```python
for grupo_a, grupo_b in combinations(columnas_asociacion, 2):
    soporte = (df_asociacion_bool[grupo_a] & df_asociacion_bool[grupo_b]).mean()

    if soporte > 0.10:
        print(grupo_a, "+", grupo_b, "=", round(soporte, 4))
```

```python
df_asociacion_bool.to_csv("Ejercicio_asociacion.csv", index=False)

print("\nDataset preparado correctamente para reglas de asociación.")
```
